## **18. Delete all entries where the status is 'inactive'.**

````python
users = {
    'u1': {'name': 'Amit', 'status': 'active'},
    'u2': {'name': 'Riya', 'status': 'inactive'},
    'u3': {'name': 'Sohan', 'status': 'active'},
    'u4': {'name': 'Neha', 'status': 'inactive'},
    'u5': {'name': 'Karan', 'status': 'active'}
}
keys_to_delete = []

for key, value in users.items():
    if value['status'] == 'inactive':
        keys_to_delete.append(key)

for key in keys_to_delete:
    del users[key]

print(users)
````

## **20. Convert all inner dictionary keys to uppercase.**

````python
students_upper = {
    'A': {'name': 'Rahul', 'age': 15, 'grade': 10},
    'B': {'name': 'Sneha', 'age': 16, 'grade': 11},
    'C': {'name': 'Arjun', 'age': 17, 'grade': 12}
}
result_dict = {}

for outer_key,details in students_upper.items():
    new_inner_dict = {}
    for key,value in details.items():
       new_inner_dict[key.upper()] = value
    
    result_dict[outer_key] = new_inner_dict

result_dict
````
