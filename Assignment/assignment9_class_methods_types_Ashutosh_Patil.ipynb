## Q1: Create a class called Student

Requirements:
- Class variable: `school_name = "ABC International School"`
- Constructor: takes `name` and `marks`
- Instance method: `display_details()`
- Class method: `change_school()`
- Static method: `is_pass(marks)` returns "Pass" if marks ≥ 35 else "Fail"


```python
class school:
    school_name="ABC International School"

    def __init__(self,name,marks):
        self.name=name
        self.marks=marks

    def display_details(self):
        print(f"name is {self.name} and marks is {self.marks}")

    @classmethod
    def change_school(cls):
        cls.school_name="Abhinav school"
        print(cls.school_name)

    @staticmethod
    def is_pass(marks):
        if marks>=35:
            return "pass"
        else:
            return "fail"

s=school("ashutosh",85)
s.display_details()
school.change_school()

result=school.is_pass(80)
result
```
