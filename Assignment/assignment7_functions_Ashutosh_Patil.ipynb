### Write an adapter function that takes a list of operations and executes add, sub, mul, div based on input. 

### Example: [ {"op": "add", "a": 10, "b": 20}, {"op": "mul", "a": 4, "b": 5} ]

```python
operation=[{"op": "add", "a": 10, "b": 20}, {"op": "mul", "a": 10, "b": 5}]

def add(a,b):
    return a+b
def mul(a,b):
    return a*b


def adapter(operation_list):
    results=[]
    for item in operation_list:
        op=item["op"]
        a=item["a"]
        b=item["b"]

        if op=="add":
            results.append(add(a,b))
        elif op=="mul":
            results.append(mul(a,b))
        else:
            print("not valid opearation")
    

    return results
   
output=adapter(operation)
print(output)
```

### Write a function process_students(data) that prints total marks and grade for each student.
- Example: [ {"name":"A", "marks":[80,90,70]}, {"name":"B", "marks":[60,75,65]} ] 
- Expected Output: Name: A | Total: 240 | Grade: A Name: B | Total: 200 | Grade: B

```python
# Example: [ {"name":"A", "marks":[80,90,70]}, {"name":"B", "marks":[60,75,65]} ] 
# - Expected Output: Name: A | Total: 240 | Grade: A Name: B | Total: 200 | Grade: B


def process_students(data):
    for student in data:
        name = student["name"]
        total = sum(student["marks"])
        
        if total >= 240:
            grade = "A"
        elif total >= 200:
            grade = "B"
        elif total >= 150:
            grade = "C"
        else:
            grade = "D"

        print(f"Name: {name} | Total: {total} | Grade: {grade}")


students = [
    {"name": "A", "marks": [80, 90, 70]},
    {"name": "B", "marks": [60, 75, 65]}
]

process_students(students)
```

### Create a function apply_discounts(products) that updates price after discount. 
- Example: [ {"name":"Laptop", "price":50000, "discount":10}, {"name":"Phone", "price":20000, "discount":5} ] 
- Expected Output: Laptop final: 45000 Phone final: 19000

```python
#  Example: [ {"name":"Laptop", "price":50000, "discount":10}, {"name":"Phone", "price":20000, "discount":5} ] 
# - Expected Output: Laptop final: 45000 Phone final: 19000

def apply_discount(data):
    for product in data:
        # print(product)
        name=product["name"]
        price=product["price"]
        discount=product["discount"]

        final_price=price-(price*discount/100)

        print(name,final_price)


product_data=[ {"name":"Laptop", "price":50000, "discount":10}, {"name":"Phone", "price":20000, "discount":5}]
apply_discount(product_data)
```

### Write a function generate_report(employees) that calculates bonus based on salary slabs. 
- Slabs: 
- salary > 50,000 → bonus = 15%, 
- salary 30,000–50,000 → bonus = 10%, 
- salary < 30,000 → bonus = 5%. 

#### Example employees = 
- [ {"id": 101, "name": "Amit", "age": 28, "department": "IT", "salary": 55000}, {"id": 102, "name": "Neha", "age": 24, "department": "HR", "salary": 40000}, {"id": 103, "name": "Raj", "age": 31, "department": "Finance", "salary": 65000}, {"id": 104, "name": "Simran", "age": 29, "department": "IT", "salary": 60000}, {"id": 105, "name": "Vijay", "age": 35, "department": "Admin", "salary": 35000} ]

```python
employee_data=[ {"id": 101, "name": "Amit", "age": 28, "department": "IT", "salary": 55000}, 
 {"id": 102, "name": "Neha", "age": 24, "department": "HR", "salary": 40000}, 
 {"id": 103, "name": "Raj", "age": 31, "department": "Finance", "salary": 65000}, 
 {"id": 104, "name": "Simran", "age": 29, "department": "IT", "salary": 60000}, 
 {"id": 105, "name": "Vijay", "age": 35, "department": "Admin", "salary": 35000} ]

def generate_report(data):
    for employee in data:
        # print(employee)
        salary=employee["salary"]
        name=employee["name"]

        bonus=0

        if salary>50000:
            bonus=(salary*15)/100

        elif salary>=30000 or salary<=50000:
            bonus=(salary*10)/100
        
        elif salary<30000:
            bonus=(salary*5)/100
            

        print(f"{name}|{salary}|{bonus}")

generate_report(employee_data)       

```

### Write a function average_salary(emp_list) that returns the average salary.

```python
employee_data=[ {"id": 101, "name": "Amit", "age": 28, "department": "IT", "salary": 55000}, 
 {"id": 102, "name": "Neha", "age": 24, "department": "HR", "salary": 40000}, 
 {"id": 103, "name": "Raj", "age": 31, "department": "Finance", "salary": 65000}, 
 {"id": 104, "name": "Simran", "age": 29, "department": "IT", "salary": 60000}, 
 {"id": 105, "name": "Vijay", "age": 35, "department": "Admin", "salary": 35000} ]

def average_salary(data):
    count=0
    total_salary=0
    for employee in data:
        count+=1
        total_salary=employee["salary"]+total_salary
        
    average_salary=total_salary/count
    return average_salary

average_salary(employee_data)
```

### Write a function salary_hike(emp_list, percent) that increases each salary by given percent.

```python
employee_data=[ {"id": 101, "name": "Amit", "age": 28, "department": "IT", "salary": 55000}, 
 {"id": 102, "name": "Neha", "age": 24, "department": "HR", "salary": 40000}, 
 {"id": 103, "name": "Raj", "age": 31, "department": "Finance", "salary": 65000}, 
 {"id": 104, "name": "Simran", "age": 29, "department": "IT", "salary": 60000}, 
 {"id": 105, "name": "Vijay", "age": 35, "department": "Admin", "salary": 35000} ]

def salary_hike(data,percent):
    for employee in data:
        
        name=employee["name"]
        salary=employee["salary"]
        salary_increment=salary+(salary*percent/100)
        print(f"{name}|{salary}|{salary_increment}")

salary_hike(employee_data,10)
```

### Write out_of_stock(items) that returns all products with stock < 20.

```python
items = [
    {"product": "Keyboard", "stock": 45},
    {"product": "Mouse", "stock": 10},
    {"product": "Monitor", "stock": 5},
    {"product": "CPU", "stock": 28},
    {"product": "Pendrive", "stock": 15},
    {"product": "Speaker", "stock": 50},
]

def out_of_stock(data):
    for item in data:
        product=item["product"]
        stock=item["stock"]
        
        if stock<20:
           print(f"out of stock{product} and stock is {stock}")

out_of_stock(items)
```

### Write total_inventory_value(items) → sum(price * stock).

```python
items = [
    {"product": "Keyboard", "price": 800, "stock": 10},
    {"product": "Mouse", "price": 400, "stock": 25},
    {"product": "Monitor", "price": 7000, "stock": 5},
    {"product": "Pendrive", "price": 600, "stock": 40},
    {"product": "CPU", "price": 15000, "stock": 3}
]

def total_inventory(items):
    sum=0
    for product in items:
        # print(product)
        price=product["price"]
        stock=product["stock"]
        sum=sum+(price*stock)

    print(sum)
total_inventory(items)
```

### Write search_product(items, name) → return matching product(s).

```python
items = [
    {"product": "Keyboard", "price": 800, "stock": 10},
    {"product": "Mouse", "price": 400, "stock": 25},
    {"product": "Monitor", "price": 7000, "stock": 5},
    {"product": "Pendrive", "price": 600, "stock": 40},
    {"product": "Mouse Pad", "price": 200, "stock": 100}
]


def search_product(items,name):
    
    for each_product in items:
        product=each_product["product"]
        if product==name:
            return print(f"product found|{product}")
       
search_product(items,"Mouse") 
```
